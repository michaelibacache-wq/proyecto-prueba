```python
import pandas as pd

# ==========================
# ARCHIVOS DE ENTRADA
# ==========================

archivo1 = "Mantenciones_pend_18112026.xlsx"
archivo2 = "Receptores en programa al 23-11-2026.xlsx"

# ==========================
# LECTURA DE ARCHIVOS
# ==========================

df1 = pd.read_excel(archivo1)
df2 = pd.read_excel(archivo2)

# ==========================
# COLUMNAS REQUERIDAS
# ==========================

columnas_df1 = [
    "Nro_ISP",
    "Muestra",
    "Estado",
    "Caja"
]

columnas_df2 = [
    "Nro_ISP",
    "ESTADOPROGRAMATX",
    "PREVISION",
    "PRAMAXIMO",
    "FECHAULTIMOPRA",
    "ULTIMOPRA"
]

df1 = df1[columnas_df1]
df2 = df2[columnas_df2]

#para normalizar tipo valor Nro_ISP

df1["Nro_ISP"] = (
    df1["Nro_ISP"]
    .astype(str)
    .str.strip()
)

df2["Nro_ISP"] = (
    df2["Nro_ISP"]
    .astype(str)
    .str.strip()
)
# ==========================
# UNIÓN DE LAS BASES
# ==========================

df_final = pd.merge(
    df1,
    df2,
    on="Nro_ISP",
    how="left"  # conserva todos los registros de Mantenciones_pend_18112026
)

# ==========================
# GUARDAR RESULTADO
# ==========================

archivo_salida = "Base_consolidada.xlsx"

df_final.to_excel(
    archivo_salida,
    index=False
)

print(f"Archivo generado: {archivo_salida}")
print(f"Registros consolidados: {len(df_final)}")
```
