```python
import pandas as pd

# ==========================================
# RUTAS DE ARCHIVOS
# ==========================================

archivo_seroteca = "BD_Seroteca_24112026.xlsx"
archivo_muestras = "Muestras ISP (1).xlsx"

# Archivo de salida
archivo_salida = "Base_Completa_ISP.xlsx"

# ==========================================
# CARGA DE DATOS
# ==========================================

df_seroteca = pd.read_excel(archivo_seroteca)
df_muestras = pd.read_excel(archivo_muestras)

# ==========================================
# LIMPIEZA DE COLUMNAS (Quitar espacios en blanco invisibles)
# ==========================================

df_seroteca.columns = df_seroteca.columns.str.strip()
df_muestras.columns = df_muestras.columns.str.strip()

# Convertir ID a texto para evitar problemas de formato
df_seroteca["Nro_ISP"] = df_seroteca["Nro_ISP"].astype(str).str.strip()
df_muestras["Nro_ISP"] = df_muestras["Nro_ISP"].astype(str).str.strip()

# ==========================================
# COLUMNAS A RESCATAR
# ==========================================

seroteca_cols = [
    "Nro_ISP",
    "Caja_ISP"
]

muestras_cols = [
    "Muestra",
    "Nro_ISP",
    "Clase_I_MFI",
    "clase_II_MFI",
    "Clase_I_Score",
    "Clase_II_Score"
]

# Mantener sólo las columnas necesarias
df_seroteca_sel = df_seroteca[seroteca_cols].copy()
df_muestras_sel = df_muestras[muestras_cols].copy()

# ==========================================
# ELIMINAR DUPLICADOS EN SEROTECA
# ==========================================

df_seroteca_sel = df_seroteca_sel.drop_duplicates(
    subset="Nro_ISP",
    keep="first"
)

# ==========================================
# MERGE
# Se conservan TODOS los Nro_ISP
# de Muestras ISP (1)
# ==========================================

df_final = df_muestras_sel.merge(
    df_seroteca_sel,
    on="Nro_ISP",
    how="left"
)

# ==========================================
# EXPORTAR
# ==========================================

df_final.to_excel(
    archivo_salida,
    index=False
)

print(f"Archivo generado exitosamente: {archivo_salida}")
print(f"Registros procesados: {len(df_final)}")
```
